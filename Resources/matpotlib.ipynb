```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
```

# Type of Plots

## Line Plots

```python
x=[1,2,3,4,5]
y=[10,20,25,30,40]
```

```python
plt.plot(x,y);
plt.title("Simple Line Plot");
plt.xlabel('X-axis label');
plt.ylabel('Y-axis label');
```

#### Bar Chart

```python
months=['Jan','Fab','Mar','Apr']
revenue=[100,400,300,500]
```

```python
plt.figure(figsize=(10,6))
plt.bar(months,revenue)
plt.title("Monthly Revenue Bar Chart");
plt.xlabel('Months');
plt.ylabel('Revenue');
```

#### Scatter plot

#### Subplots and Legends

```python
days=[1,2,3,4,5]
original_price=[10,20,25,30,40]
predicted_price=[8,21,25,28,38]
```

```python
plt.plot(days, original_price, 'b-o', label='Original Price')
plt.plot(days, predicted_price, 'r-o',label='Predicted price')
plt.legend()
plt.title('Plot with legends')
plt.xlabel('Weekdays')
plt.ylabel('Price')
```

### Subplots

```python
# First subplot - original Price
plt.subplot(3,1,1)
plt.plot(days, original_price, 'b-o', label='Original Price')
plt.title('Original prices over time')
plt.xlabel('weekdays')
plt.ylabel('Price')
plt.legend()


# second subplot - Predicted Price
plt.subplot(3,1,2)
plt.plot(days, predicted_price, 'r-o',label='Predicted price')
plt.title('Predicted Prices over time')
plt.xlabel('weekdays')
plt.ylabel('Price')
plt.legend()


# Third subplot - Difference Between Original & Predicted Price
Price_difference=[o - p for o,p in zip(original_price, predicted_price)]
plt.subplot(3,1,3)
plt.bar(days, Price_difference, color='purple',label='price Difference')
plt.title('Difference Between Original & Predicted Price')
plt.xlabel('Days')
plt.ylabel('Price Difference')
plt.legend()


plt.tight_layout()
```

#### Saving the figure

```python
plt.plot(days, original_price, 'b-o', label='Original Price')
plt.plot(days, predicted_price, 'r-o',label='Predicted price')
plt.legend()
plt.title('Plot with legends')
plt.xlabel('Weekdays')
plt.ylabel('Price')
plt.savefig('Example_prediction.png', dpi=300) #save the image
```
